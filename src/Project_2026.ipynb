```python
import pandas as pd
import numpy as np

```

```python
file_path = '../Data/raw_data.csv'
df = pd.read_csv(file_path, encoding='cp1255')

```

```python
info_df = pd.DataFrame({
    'Column Name': df.columns,
    'Data Type': df.dtypes.values,
    'Missing Values': df.isnull().sum().values,
    'Missing (%)': (df.isnull().sum().values / len(df) * 100).round(1),
    'Unique Values': df.nunique().values
})
display(info_df)
```

```python
df_new = df.iloc[:, [2, *range(6, 26)]].copy()

df_new.columns = [
    'Submission_Time',
    'Agreement_2020',
    'Has_Covid_Clause',
    'Covid_Clause_Details',
    'Household_Income_Affected',
    'Monthly_Rent',
    'Rent_Changed',
    'Requested_Rent_Reduction',
    'Landlord_Agreed_Rent_Reduction',
    'Requested_Payment_Installments',
    'Requested_Shorter_Lease',
    'Lease_Terminated_Early',
    'Eviction_Action',
    'Rental_Area',
    'Marital_Status',
    'Age',
    'Gender',
    'Children_Living_At_Home',
    'Income_Relative_To_Average',
    'Employment_Status',
    'Suggested_Solution'
]

df_new
```

### creating cleand columns ###

```python
agreement_mapping = {
    'לפני': 0,
    'אחרי': 1
}

df_new['Agreement_2020'] = df_new['Agreement_2020'].astype(str).str.strip().map(agreement_mapping).astype('Int64')
```

```python
yes_no_mapping = {
    'כן': 1,
    'לא': 0
}

binary_cols = [
    'Has_Covid_Clause',
    'Household_Income_Affected',
    'Rent_Changed',
    'Requested_Rent_Reduction',
    'Landlord_Agreed_Rent_Reduction',
    'Requested_Shorter_Lease',
    'Eviction_Action',
    'Children_Living_At_Home'
]

for col in binary_cols:
    df_new[col] = (df_new[col].astype(str).str.strip().map(yes_no_mapping).astype('Int64'))
```

```python
rent_mapping = {
    'עד 3000 ₪ בחודש': 0,
    'בין 3001 ל-5000 ₪ בחודש': 1,
    'בין 5001 ל-7000 ₪ בחודש': 2,
    'מעל 7001 ₪ בחודש': 3
}

df_new['Monthly_Rent'] = (df_new['Monthly_Rent'].astype(str).str.strip().map(rent_mapping).astype('Int64'))
```

```python
income_mapping = {
    'הרבה מתחת הממוצע': -2,
    'מעט מתחת לממוצע': -1,
    'כמו הממוצע': 0,
    'מעט מעל הממוצע': 1,
    'הרבה מעל הממוצע': 2
}

df_new['Income_Relative_To_Average'] = (df_new['Income_Relative_To_Average'].astype(str).str.strip().map(income_mapping).astype('Int64'))
```

```python
gender_mapping = {
    'גבר': 0,
    'אישה': 1
}

df_new['Gender'] = (df_new['Gender'].astype(str).str.strip().map(gender_mapping).astype('Int64'))

```

```python
### CODE BY GEMINI ###
# 1. מיפוי לעמודה החדשה: האם המשכיר הסכים לפריסת תשלומים?
agreed_installments_mapping = {
    'כן, והמשכיר הסכים': 1,
    'כן, אך המשכיר לא הסכים': 0
}

agreed_col = (
    df_new['Requested_Payment_Installments']
    .astype(str)
    .str.strip()
    .map(agreed_installments_mapping)
    .astype('Int64')
)

# מציאת המיקום של העמודה המקורית והכנסת העמודה החדשה מיד אחריה (+1)
col_idx = df_new.columns.get_loc('Requested_Payment_Installments') + 1
df_new.insert(loc=col_idx, column='Landlord_Agreed_Installments', value=agreed_col)

# 2. מיפוי לעמודה המקורית: האם השוכר ביקש פריסת תשלומים? (0 = לא, 1 = כן)
requested_installments_mapping = {
    'לא ביקשתי': 0,
    'כן, אך המשכיר לא הסכים': 1,
    'כן, והמשכיר הסכים': 1
}

df_new['Requested_Payment_Installments'] = (
    df_new['Requested_Payment_Installments']
    .astype(str)
    .str.strip()
    .map(requested_installments_mapping)
    .astype('Int64')
)
```

```python
df_new['Covid_Clause_Details'] = df_new['Covid_Clause_Details'].str.strip()

no_clause_values = [
    'אין הסדר',
    'אין',
    'אין סעיף שקשור לקורונה.',
    'אין לי כזה סעיף',
    'אין הסדר כזה',
    'אין סעיף כזה',
    'אין התייחסות',
    'אין הסכם'
]

df_new['Covid_Clause_Details'] = df_new['Covid_Clause_Details'].replace(
    no_clause_values, 
    'לא קיים הסדר'
)
```
